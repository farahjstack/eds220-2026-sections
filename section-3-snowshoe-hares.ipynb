```python
import pandas as pd
import numpy as np

hares = pd.read_csv("data/55_Hare_Data_2012.txt")
```

```python
# View first few rows
hares.head()
```

```python
# Check dimensions
hares.shape
```

```python
# Column types and non-null counts
hares.info()
```

```python
# Missing values per column
hares.isna()
```

```python
# Sums all of boolean columns
hares.isna().sum()
```

```python
# How many rows vs unique hares?
print(hares.shape[0], hares['l_ear'].nunique())

# Details: hares.shape[0] returns the number of rows in the DataFrame - the total count of capture events
# hares['l_ear'].unique(())
```

```python
# Recapture the distrubution (how many values exist for each of the hare we have)
hares.groupby('l_ear').size()

# Define into a variable
captures_per_hare = hares.groupby('l_ear').size()

captures_per_hare.describe()
```

```python
# Min and max for continuous measurements
print("Weight (g):", hares['weight'].min(), "", hares['weight'].max())
```

```python
# Unique values for key categorical columns
print("grid:", hares['grid'].unique())
print("age:", hares['age'].unique())
print("study:", hares['study'].unique())
print("sex:", hares['sex'].unique())
```
